```python
if spark.catalog.tableExists("data_modelling.bronze.source_data"):
    last_load_data = spark.sql("select max(order_date) from data_modelling.default.source_data").collect()[0][0] 
else: 
    last_load_data = '2023-12-31'
```

```python
spark.sql(f"""
select * from data_modelling.default.source_data
where order_date > '{last_load_data}'
""").createOrReplaceTempView("bronze_source")
```

```python
%sql
create or replace table data_modelling.bronze.bronze_table
AS
select * from bronze_source
```
